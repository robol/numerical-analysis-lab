### 1.2. Il metodo di Jacobi

Per risolvere $Ax=b$, il metodo di Jacobi usa

$$x_i^{(\mathrm{new})}=\frac{1}{A_{ii}}\left(b_i-\sum_{j=1}^{i-1}A_{ij}x_j^{(\mathrm{old})}-\sum_{j=i+1}^n A_{ij}x_j^{(\mathrm{old})}\right),\qquad i=1,\ldots,n.$$

Tutte le componenti nuove dipendono da quelle **vecchie**. Non sovrascrivete `x[i]` prima di aver completato l’intero passo. In Julia `x_new = x` non copia il vettore: allocate `x_new = zeros(n)` oppure usate `copy(x)`.

### 1.3. Il metodo di Gauss–Seidel

La formula è

$$x_i^{(\mathrm{new})}=\frac{1}{A_{ii}}\left(b_i-\sum_{j=1}^{i-1}A_{ij}x_j^{(\mathrm{new})}-\sum_{j=i+1}^n A_{ij}x_j^{(\mathrm{old})}\right).$$

L'unica differenza è che nella prima somma si usano le componenti già aggiornate.

### Esercizio 4

Quale dei due metodi è implementato nella funzione seguente? Perché? `!=` significa “diverso”.

```julia
function mystery(A, b, k)
    n = size(A, 1)
    x = float.(b)
    for it = 1:k
        for i = 1:n
            t = b[i]
            for j = 1:n
                if i != j
                    t = t - A[i, j] * x[j]
                end
            end
            x[i] = t / A[i, i]
        end
        println(norm(A*x-b))
    end
    return x
end
```

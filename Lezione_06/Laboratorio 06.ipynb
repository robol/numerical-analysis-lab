### 1.2. Il metodo di Jacobi

La formula che definisce il metodo di Jacobi per la risoluzione del sistema $Ax=b$ è

$$x_i^{(\mathrm{new})}=\frac{1}{A_{ii}}\left(b_i-\sum_{j=1}^{i-1}A_{ij}x_j^{(\mathrm{old})}-\sum_{j=i+1}^n A_{ij}x_j^{(\mathrm{old})}\right),\qquad i=1,\ldots,n.$$

### 1.3. Il metodo di Gauss–Seidel

La formula che definisce il metodo di Gauss--Seidel per la risoluzione del sistema $Ax=b$ è

$$x_i^{(\mathrm{new})}=\frac{1}{A_{ii}}\left(b_i-\sum_{j=1}^{i-1}A_{ij}x_j^{(\mathrm{new})}-\sum_{j=i+1}^n A_{ij}x_j^{(\mathrm{old})}\right),\qquad i=1,\ldots,n.$$

Si noti che l'unica differenza rispetto al metodo di Jacobi è che nella prima somma si usano le componenti del vettore aggiornato.

### Esercizio 4

Quale dei due metodi è implementato nella funzione seguente? Perché?

```julia
function mystery(A, b, k)
# esegue k passi di uno tra i metodi di Jacobi e Gauss-Seidel: quale?
    n = size(A, 1)
    x = float.(b) # o qualunque altro vettore iniziale
    for it = 1:k
        for i = 1:n
            t = b[i] # accumulatore
            for j = 1:n
                if i != j # ricordate che != e' il simbolo di diverso
                    t = t - A[i, j] * x[j]
                end
            end
            x[i] = t / A[i, i]
        end
        println(norm(A*x-b)) # stampa la norma del residuo
    end
    return x
end
```
